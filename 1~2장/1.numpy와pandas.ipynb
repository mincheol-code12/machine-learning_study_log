```python
# numpy
import numpy as np

# array? : 머신러닝 알고리즘에서 요구하는 입출력 데이터 형태
array1 = np.array([1,2,3])
print(type(array1))
print(array1.shape)

array2 = np.array([[1,2,3],[4,5,6]])
print(array2.shape)

array3 = np.array([[1,2,3]])
print(array3.shape)

# array 1과 array3 는 형태는 비슷하지만 각각 1차원과 2차원이라는 차이가 있음, array1은 단순히 숫자가 들어간 배열일뿐, 원소 위치를 찾을때 숫자 1개면되지만
# array3은 행렬이기때문에 인덱스가 2개필요함

print(array1.ndim,array3.ndim,array2.ndim)
```

```python
# list를 쉽게 array로 바꿀수있음 
list1=[1,2,3]
print(type(list1))
array1=np.array(list1)
print(array1,array1.dtype)

# ndarray내의 데이터 타입은 연산의 특성상 같은 데이터 타입만 가능한대, 만약에 다른게 섞여있으면 크기가 더큰 데이터 타입으로 형변환을 일괄적용

list2=[1,2,'test']
array2=np.array(list2)
print(array2,array2.dtype)  # < 1,2과 문자타입으러

list3=[1,2,3.0]
array3=np.array(list3)
print(array3,array3.dtype) #<1,2가 float으로
```

```python
# 메모리줄이기  astype으로 더 작은 데이터 타입으로 바꿔 계산하는게 좋음, 메모리 문제 때문에

array_int=np.array([1,2,3])
array_float=array_int.astype('float64')
print(array_float,array_float.dtype)


array_int1=array_float.astype('int32')
print(array_int1,array_int1.dtype)

array_float=np.array([1.2,2.3,3.3])
array_int2=array_float.astype('int32')
print(array_int2,array_int2.dtype)
print(array_float.shape)
print(array_int2.shape)
```

```python
# ndarray 편리하게 생성하기  arange/zeros/ones
# test데이터를 만들거나,대규모 데이터를 일괄적으로 초기화해야할때 사용함


#arrange : range에해당하는 범위를 array로 만드는거 , (start,end,몇씩차이나게할건지?)
sequence_array=np.arange(10)
print(sequence_array,sequence_array.shape)

#zeros(): 튜플 형태의 shape값을 입력하면 모둔 값을 0으로 채운 shape를 가진 array를 반환
zero_array=np.zeros((3,2),dtype='int32')
print(zero_array,zero_array.dtype)

#ones(): zeros와 동일 얘는 1로 채움
one_array=np.ones((3,2),dtype='int32') # dtype입력안하면 int64로채움
print(one_array,one_array.dtype)
```

```python
# ndarray의 차원과 크기를 변경하는 reshape()

array=np.arange(10)
print(array,array.shape)

array2=array.reshape(1,10)
print(array2,array2.shape)

array3=array.reshape(2,5)
print(array3,array3.shape)

# reshape에 -1을 사용하면 ndarray에 호환되는 shape로 자동변환해줌 
array4=array.reshape(5,-1)
print(array4,array.shape)

#자주 쓰는 형태 reshape(-1,1) 무조건 2차원으로 차원수를 바꿔버림
array5=array.reshape(1,-1)
print(array5,array5.shape)

array1=np.arange(8)
array3d=array1.reshape(2,2,2) # depth,행,렬 > 행렬 묶음이 2개 
print(array3d.tolist())

# 3차원 ndarray를 2차원으로
array5=array3d.reshape(-1,1)
print(array5.tolist(),array5.shape)

# 1차원을 2차원으로
array6 =array1.reshape(-1,1)
print(array6.tolist(),array6.shape)
```

```python
# numpy -indexing

#1.특정 데이터만 추출하기
array1= np.arange(1,10)
print(array1,array1[2],array[-1],array[-2])
array1[2]=15
print(array1)

array1d=np.arange(1,10)
array2d=array1d.reshape(3,3)

print(array2d)
print(array2d[0,0],array2d[2,2],array2d[1,2])
# axis=0은 -> 행방향, axis=1은 열방향 3차원의 경우 axis=2는 높이방향

#2.슬라이싱
array1=np.arange(1,10)
print(array1[0:3]) # 단일 추출을 제외한 나머지 인덱싱은 모두 데이터타입이 ndarray임
print(array1[:3])
print(array1[3:])

array1d=np.arange(1,10)
array2d=array1d.reshape(3,3)
print(array2d[0:2,0:2]) # 2차원이니까 0~1행까지 , 0열~1열까지
print(array2d[1:3,0:3]) # 1행~2행 , 0열~2열까지
print(array2d[1:3,:])
print(array2d[:2,1:]) # 0~1행 , 1~2열
print(array2d[:2,0])
# 2차원 or n차원 ndarray에서 뒤에오는 인덱스를 지우면  n-1차원 ndarray를 반환
print(array2d[0]) # axis=0 row방향값 전부를 보여줌 array2d[0,:] 동일


#3.팬시 인덱싱 : 일정한 인덱싱 집합을 리스트또는 ndarray로 지정해 해당위치에 있는 데이터의 ndarray를 반환
array3= array2d[[0,1],2] # array2d[0,2], array2d[1,2]
print('팬시',array2d,array3.tolist()) 

array4=array2d[[0,1],0:2]
print('팬시2',array2d,array4.tolist())

#4. 불린 인덱싱 : 특정 조건에 해당하는지 여부인 true/false값을 기반으로 true에 해당하는 인덱스에 있는 위치 데이터의 ndarray를 반환

array1d=np.arange(1,10)
array3=array1d[array1d>5]
print('불린',array3)
print(array1d>5)

불린_인덱스=np.array([False,False,False,False,False,True,True,True,True])
print(array1d[불린_인덱스])
a='False False False False False  True  True  True  True'
mask = np.array([x == 'True' for x in a.split()])
print(array1d[mask])
```

```python
# 행렬의 정렬 sort()와 argsort()
# np.sort : 원행렬을 유지한체 정렬된행렬을 반환 , ndarray.sort() : 원행렬 자체를 정렬 변환 

org_array =np.array([3,1,9,5])
sort_array1=np.sort(org_array)
print(sort_array1)

sort_array2 =org_array.sort()
print('반환행렬', sort_array2) 
print('원본행렬',org_array)

# 내림차순 정렬
sort_array1_desc=np.sort(org_array)[::-1]
print(sort_array1_desc)

# axis=0은 행방향향 아래 0행부터~n행순서 아래화살표 , axis=1은 칼럼방향 오른쪽  오른쪽 화살표 

array2d=np.array([[8,12],[7,1]])

sort_array2d_axis0=np.sort(array2d,axis=0)
sort_array2d_axis1=np.sort(array2d,axis=1)
print(array2d)
print('axis=0',sort_array2d_axis0)
print('axis=1',sort_array2d_axis1)

#정렬된 행렬의 인덱스 반환하기 : 원본 행렬이 정렬되었을때, 기존 원본 행렬의 원소에 대한 인덱스를 필요로할때는 np.argsort()를 사용함
org_array=np.array([3,1,8,5])
sort_인덱스=np.argsort(org_array)
print(sort_인덱스)
```

```python
# argsort의 활용방법 
# 학생이름, 학생 점수라는 두개의 배열이 존재할때, 학생 점수 순으로 학생이름을 불러낼려면?
score_array=np.array([10,5,2,7,11])
name_array=np.array(['김','이','박','최','상'])

score_array_index=np.argsort(score_array)
print(score_array_index)
name_array[score_array_index]
```

```python
# 선형대수 계산 : 행렬 내적 (행렬 곱)

A=np.array([ [1,2,3],[4,5,7]])
b=np.array([[7,8],[9,10],[11,12]])

dot_product=np.dot(A,b)
print(dot_product)
```

```python
lambda_square=lambda x:x**2
print(lambda_square(3))
```

```python
#설명

* lambda x : x**2 에서 입력인자와 계산식을 :를 통해 분리함  :의 왼쪽이 입력값 , 오른쪽이 입력 식

* 입력값으로 여러개의 값을 쓸때는 map()함수를 결합한다

a=[1,2,3]
square=map(lambda x : x**2,a)
```

```python
# apply,lambda를 이용해서 데이터프레임에 식을 적용하기 apply가 시리즈

titanic_df['Name_len']=titanic_df['Name'].apply(lambda x : len(x))
titanic_df[['Name','Name_len']].head(3)
```
