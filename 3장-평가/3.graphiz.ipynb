```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

dt_clf=DecisionTreeClassifier(random_state=42)
df=load_iris()

X_train,X_test,y_train,y_test=train_test_split(df.data,df.target,test_size=0.2,random_state=42)

dt_clf.fit(X_train,y_train)
```

```python
#tree.dot을 생서앟여 graphviz로 열수있음
from sklearn.tree import export_graphviz

export_graphviz(dt_clf,out_file='tree.dot',class_names=df.target_names,feature_names=df.feature_names,impurity=True,filled=True)
```

```python
import graphviz

with open('tree.dot') as f:
    dot_graph=f.read()
graphviz.Source(dot_graph)
```

```python
import seaborn as sns
import numpy as np

print(f'feture importance{dt_clf.feature_importances_}')

for name,value in zip(df.feature_names,dt_clf.feature_importances_):
    print(f'{name},{value}')

sns.barplot(x=dt_clf.feature_importances_,y=df.feature_names)    
```
